```python
# Exercise 1: Check the positive and negative numbers

def check_number(number):
    if number > 0:
        print("Positive")
    elif number < 0:
        print("Negative")
    else:
        print("Zero")

check_number(4)
check_number(0)
check_number(-10)
```

```python
# Area of a square

def area_of_square(side):
    area = side * side
    print(area)

area_of_square(5) 

#Using the math module

import math

import math

def area_of_square(side):
    area = math.pow(side, 2)
    print(area)

area_of_square(4)

#pow means power - power means raising a number to an exponent.
```

```python
# Area of a rectangle

def area_of_rectangle(length, width):
    area = length * width
    print(area)

area_of_rectangle(8, 4)  

#Using the math module

import math

def area_of_rectangle(length, width):
    area = math.prod([length, width])
    print(area)

area_of_rectangle(6, 3)

#prod means product (multiplication).
```

```python
# Area of a circle

def area_of_circle(radius):
    area = 3.14 * radius * radius
    print(area)

area_of_circle(5) 

#Using the math module

import math

def area_of_circle(radius):
    area = math.pi * math.pow(radius, 2)
    print(round(area))

area_of_circle(6)
```

```python
# Calculate Grades

def calculate_grade(score):
    
    if score >= 90:
        print("A")
    elif score >= 80:
        print("B")
    elif score >= 70:
        print("C")
    elif score >= 60:
        print("D")
    else:
        print("F")

calculate_grade(90)
calculate_grade(87)
calculate_grade(79)
calculate_grade(65)
calculate_grade(30)        
```
