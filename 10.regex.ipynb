## Regular Expressions

`Regualar Expressions` are patterns used to search, match, extract, validate and also manupulate text

Instead of searching for one extact value, Regex allows us to describe a different text values can follow

Python provides a string methods such as `replace()`, `split()`, `find()` and the `in` operator

The main advantage of `regex`: it allows us to work with patterns rather than individual text values

### REGEX (REGULAR EXPRESSIONS) - CHARACTER CHEAT SHEET

| Character | Meaning | Example | Matches |
|---|---|---|---|
| . | Matches any character except a newline | a.c | abc, a1c, a@c |
| ^ | Matches the beginning of a string | ^Hello | Hello world |
| $ | Matches the end of a string | world$ | Hello world |
| * | Matches 0 or more repetitions of the previous character | ab*c | ac, abc, abbc |
| + | Matches 1 or more repetitions of the previous character | ab+c | abc, abbc |
| ? | Matches 0 or 1 repetition of the previous character | ab?c | ac, abc |
| {n} | Matches exactly n repetitions of the previous character | a{3} | aaa |
| {n,m} | Matches between n and m repetitions of the previous character | a{2,4} | aa, aaa, aaaa |
| [] | Matches any one character inside the brackets | [abc] | a, b, or c |
| [^] | Matches any character NOT inside the brackets | [^abc] | d, x, 5, etc. |
| \ | Escapes a special character or introduces a special sequence | \. | A literal dot (.) |
| | | | |
| \d | Matches any digit (0-9) | \d | 7, 0, 3 |
| \D | Matches any character that is NOT a digit | \D | A, #, b |
| \w | Matches a word character (letters, digits, underscore) | \w | a, 5, _ |
| \W | Matches a character that is NOT a word character | \W | @, !, space |
| \s | Matches a whitespace character (space, tab, newline) | \s | space, tab |
| \S | Matches a character that is NOT whitespace | \S | A, 7, @ |
| | | | |
| ( ) | Groups characters together | (abc) | abc as one group |
| | | | |
| \b | Matches a word boundary | \bcat | cat at the beginning of a word |
| \B | Matches a position that is NOT a word boundary | \Bcat | cat inside a larger word |

### The Dot `(.)`

The dot matches any single character except a new line

```python
text = "cat, cot, caat, cut, bat, mat"

result = re.findall(r"c.t", text)

print(result)
```

### Charcter class

`Charcter classes` allows us to specify which characters should be matched. 

They are usually written using square brackets `[]`

```python
text = "cat, cot, caat, cut, bat, mat"

result = re.findall(r"c[ao]t", text)
print(result)
```

### `re.match()` 

Checks for a pattern only at the beginning of a string

```python
refrence = "INV-2048 REF-2078 INV-456"

result = re.match(r"INV", refrence)

#result = re.search(r"INV", refrence)

print(result.group())
```

### `re.fullmatch()`

Checks whether the entire string follows the specified pattern

```python
product_code = "AB2048"

result = re.fullmatch(r"[A-Z]{2}\d{4}", product_code)

print(result)
```
