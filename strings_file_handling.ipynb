## File Handling

### Why Do We Need File Handling?

File handling allows a python program to:

+ Read information that has already been saved 

+ Create new files

+ Write information into the files

+ Add new information without deleting existing content

### Opening A File

Python uses the `open()` function to open a file so that we can read and write information

**Syntax:**

```python
file = open("filename.txt", "mode)
```

The `mode` determines how we interact with the file

Mode	Meaning


### Common File Modes 

| Mode | Meaning |
|------|---------|
| `"r"` | Read a file |
| `"w"` | Write to a file (overwrites existing content) |
| `"a"` | Append new content to the end |
| `"x"` | Create a new file; errors if it already exists |

`NB` Be careful when using `w` because it replaces the existing content

## Wring to a file

We use `w` mode to write information into a file

````python
#student_name = input("Enter student name: ").strip().title()


with open("students.txt", "w") as file:
    file.write("Wayne Murithi\n")
    file.write("Mark Kimani\n")
    file.write("Jane Gachari\n")
````

````python
student_name = input("Enter student name: ").strip().title()
with open("student.txt", "w") as file:
    file.write(student_name)
    print("Student registered succesfully")
````

### Appending to a file

We use `a` mode to add content to a file without deleting existing records

````python
student_name = input("Enter student name: ").strip().title()
with open("student.txt", "a") as file:
    file.write(student_name)
    print("Student registered succesfully")
````

### Reading From A File

We use `r` mode to retrieve information from an existing file

````python
with open("student.txt", "r") as file:
    content = file.read()
    print(content)
````

### DIFFERENT WAYS TO READ A FILE

| Method | Meaning | Example |
|---|---|---|
| read() | Reads the entire file | file.read() |
| readline() | Reads one line at a time | file.readline() |
| readlines() | Reads all lines and stores them in a list | file.readlines() |
| for loop | Reads the file line by line | for line in file: |



````python
# Read the first_line

with open("students.txt", "r") as file:
    print(file.readline())
````

````python
# Read all lines

with open("students.txt", "r") as file:
    print(file.readlines())
````

### Reading a file line by line

We can use a loop to read the records individually

````python
with open("students.txt", "r") as file:
    for student_name in file:
        print(student_name.strip())

# We used .strip to get rid of the spaces
````

### Closing a file 

````python
file = open("students.txt", "r")
content = file.read()
print(content)
file.close()
````

### Using `with open()` 

The recommended way to work with files 

The file is automatically closed

````python
with open("students.txt", "x") as file:
    content = file.read()
    print(content)
````
